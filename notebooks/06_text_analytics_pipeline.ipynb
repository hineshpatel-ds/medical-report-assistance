```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
import os

# load preprocessed clincial notes
df = pd.read_csv("../data/processed/clinical_notes_preprocessed.csv")

# Initialize Tfidf Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=500, stop_words="english", ngram_range=(1,2))

#fit and transform
X_tfidf = tfidf_vectorizer.fit_transform(df['processed_text'])

# create dataframe
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

# save for next steps
os.makedirs("../data/features", exist_ok=True)
tfidf_df.to_csv("../data/features/clinical_notes_tfidf.csv",index=False)

print("TF-IDF representation created and saved at '../data/features/clinical_notes_tfidf.csv'")
print(tfidf_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#compute mean TF-IDF score for each term
mean_tfidf = tfidf_df.mean().sort_values(ascending=False).head(20)

plt.figure(figsize=(10,6))
sns.barplot(x=mean_tfidf.values, y=mean_tfidf.index, palette="viridis")
plt.title("Top 20 words by mean TF-IDF score", fontsize=14)
plt.xlabel("Mean TF_IDF weight")
plt.ylabel("Term")
plt.show()

```
